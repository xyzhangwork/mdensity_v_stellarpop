```python
import numpy as np
from astropy.io import fits
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import matplotlib as mpl
import sys
from read_alf import Alf
import warnings
from matplotlib.legend_handler import HandlerTuple

warnings.filterwarnings("ignore")
```

```python
parameters = fits.open('../data/parameters_reff2_2.fits')
split = fits.open('../data/sample_split_outliers_total2.fits')
rnames = ['rin_median(kpc)', 'rmid_median(kpc)', 'rout_median(kpc)']
split_type = ['m20kpc_v_sigmacen', 'm10kpc_v_m20kpc']
split_num = 1
label_names = ['high', 'low']
radius_name = ['in', 'mid', 'out']
labels = ['High', 'Low']

ele_abundances = ['Mg','a','N','C','Ca','Na']
params = ['Mg','a','N','C','Ca','Na']

keywords = ['FeH', 'Mg', 'Mg/Fe','a', 'a/Fe',
            'N','N/Fe','C','C/Fe','Na','Na/Fe',
            'Ca','Ca/Fe','r_re', 'r_kpc','Age','zh']

path = '../data/alf/simple/'
data = {}
err = {}
sigdata = {}

for i in range(2):
    for name in keywords:
        data[name] = [[[], [], []], [[], [], []]]
        err[name] = [[[], [], []], [[], [], []]]
        sigdata[name] = [[], [], []]

for j in range(3):
    for i in range(2):
        myalf = Alf(
            path + split_type[split_num] + '/' + label_names[i] + '_' + split_type[split_num] + '_'
            + radius_name[j])
        myalf.get_total_met()
        myalf.abundance_correct()
        zh = np.where(myalf.labels == 'zH')
        mg = np.where(myalf.labels == 'Mg')
        feh = np.where(myalf.labels == 'FeH')

        data['FeH'][i][j].append(myalf.tmet['cl50'])
        err['FeH'][i][j].append(myalf.tmet['std'])
        data['Age'][i][j].append(myalf.results['logage'][0])
        err['Age'][i][j].append(myalf.results['logage'][2])
        z_mcmc=myalf.mcmc[:, feh].flatten()+myalf.mcmc[:, zh].flatten()+0.94*myalf.get_corrected_abundance_posterior('Mg')
        z_data = myalf.get_cls(z_mcmc)
        data['zh'][i][j].append(z_data['cl50'])
        err['zh'][i][j].append(z_data['std'])

        for name in params:
            para_mcmc = myalf.mcmc[:, feh].flatten()+myalf.mcmc[:, zh].flatten()+myalf.get_corrected_abundance_posterior(name)
            para_tmp=myalf.get_cls(para_mcmc)
            data[name][i][j].append(para_tmp['cl50'])
            err[name][i][j].append(para_tmp['std'])
        for name in ele_abundances:
            data[name + '/Fe'][i][j].append(myalf.xFe[name]['cl50'])
            err[name + '/Fe'][i][j].append(myalf.xFe[name]['std'])
fdata = {}
ferr = {}
fsigdata = {}

for i in range(2):
    for name in keywords:
        fdata[name] = [[[], [], []], [[], [], []]]
        ferr[name] = [[[], [], []], [[], [], []]]
        fsigdata[name] = [[], [], []]

comp_path = '../data/alf/full/'
for j in range(3):
    for i in range(2):
        myalf = Alf(
            comp_path + split_type[split_num] + '/' + label_names[i] + '_' + split_type[split_num] + '_'
            + radius_name[j]+'_full')
        myalf.get_total_met()
        myalf.abundance_correct()
        zh = np.where(myalf.labels == 'zH')
        mg = np.where(myalf.labels == 'Mg')
        feh = np.where(myalf.labels == 'FeH')

        fdata['FeH'][i][j].append(myalf.tmet['cl50'])
        ferr['FeH'][i][j].append(myalf.tmet['std'])
        fdata['Age'][i][j].append(myalf.results['logage'][0])
        ferr['Age'][i][j].append(myalf.results['logage'][2])
        z_mcmc = myalf.mcmc[:, feh].flatten() + myalf.mcmc[:,
                                                zh].flatten() + 0.94 * myalf.get_corrected_abundance_posterior('Mg')
        z_data = myalf.get_cls(z_mcmc)
        fdata['zh'][i][j].append(z_data['cl50'])
        ferr['zh'][i][j].append(z_data['std'])

        for name in params:
            para_mcmc = myalf.mcmc[:, feh].flatten() + myalf.mcmc[:,
                                                       zh].flatten() + myalf.get_corrected_abundance_posterior(name)
            para_tmp = myalf.get_cls(para_mcmc)
            fdata[name][i][j].append(para_tmp['cl50'])
            ferr[name][i][j].append(para_tmp['std'])
        for name in ele_abundances:
            fdata[name + '/Fe'][i][j].append(myalf.xFe[name]['cl50'])
            ferr[name + '/Fe'][i][j].append(myalf.xFe[name]['std'])
```

```python
plt.rcParams.update({
    "text.usetex": True,
    "font.family": "Charter",
    "xtick.labelsize": 19,
    "ytick.labelsize": 19,
    "axes.labelsize": 21,
    "axes.labelweight": "bold"
    })
title_s = ['FeH','Mg/Fe','Age']
title_l = ['a/Fe', 'N/Fe','C/Fe','Ca/Fe','Na/Fe']
sga_mass = fits.getdata('../data/sga_mass_new_28mag2.fits',1)
title_s = np.array(title_s)
title_l=np.array(title_l)
title_s0 = np.array([r'$\mathbf{[Fe/H]}$', r'$\mathbf{[Mg/Fe]}$',r'$\mathbf{log\ Age(Gyr)}$'])
title_l0=np.array([r'$\mathbf{[O/Fe]}$', r'$\mathbf{[N/Fe]}$', r'$\mathbf{[C/Fe]}$',r'$\mathbf{[Ca/Fe]}$',r'$\mathbf{[Na/Fe]}$'])
normalize = mcolors.Normalize(vmin=-1.3, vmax=0.3)
colormap = mpl.cm.PRGn
color_low = colormap(normalize(-1))
color_high = colormap(normalize(0))
colors = [color_high, color_low]
edgecolors = ['xkcd:forest', 'xkcd:blue violet']
shapes = ['o', 'd']
sample_labels = [r'$\mathrm{Extended}$', r'$\mathrm{Compact}$']

if split_num == 0:
    colors = ['xkcd:tangerine', 'xkcd:dark sky blue']
    edgecolors = ['xkcd:copper', 'xkcd:vivid blue']
    sample_labels = [r'$\mathrm{Higher\ \sigma}$', r'$\mathrm{Lower\ \sigma}$']
    shapes = ['p', '*']
fig_s, ax_s = plt.subplots(1, len(title_s0), figsize=(4.4 * len(title_s0), 4), clear=True)
for idx_cal, name in enumerate(title_s):
    func = []
    ax_s[idx_cal].set_title(title_s0[idx_cal], x=.5, y=1.04, fontsize=23)
    ax_s[idx_cal].set_xlabel(r'$R\mathrm{/kpc}$')
    xmin, xmax = 10, 10
    for i in range(2):
        mask = split[1].data[split_type[split_num]] % 2 == i
        ifus, idx, _ = np.intersect1d(parameters[1].data['plateifu'], split[1].data['plateifu'][mask],
                                      return_indices=True)
        _, idx_re, idx0 = np.intersect1d(sga_mass['plateifu'], ifus, return_indices=True)
        re = sga_mass['Re_kpc'][idx_re]
        dx_tmp, dy_tmp = [], []
        for j in range(3):
            # r_tmp=0.5 * j + i * 0.04
            r_tmp = np.nanmedian(parameters[1].data[rnames[j]][idx])
            re_tmp = np.nanmedian(parameters[1].data[rnames[j]][idx] / re)
            data['r_kpc'][i][j].append(r_tmp)
            data['r_re'][i][j].append(re_tmp)

            # idx_r = np.where(r_median[1].data['plateifu'] == '10513-12701')[0][0]
            if r_tmp < xmin:
                xmin = np.copy(r_tmp)
            if r_tmp > xmax:
                xmax = np.copy(r_tmp)

            ax_s[idx_cal].errorbar([r_tmp], [data[name][i][j]],
                                   xerr=[np.nanstd(parameters[1].data[rnames[j]][idx])],
                                   yerr=[err[name][i][j]],
                                   ecolor=edgecolors[i], linewidth=1.5, capsize=10, capthick=2, linestyle=None,
                                   alpha=1)
            ax_s[idx_cal].scatter([r_tmp], [data[name][i][j]], s=200,
                                  edgecolors=edgecolors[i], facecolors=colors[i], linewidths=1.5,
                                  marker=shapes[i],
                                  alpha=1)
            dx_tmp.append(r_tmp)
            dy_tmp.append(data[name][i][j][0])

s1 = ax_s[-1].scatter([], [], s=200, marker=shapes[0], facecolor=colors[0], edgecolor=edgecolors[0],
                      label=sample_labels[0], lw=2)
s2 = ax_s[-1].scatter([], [], s=200, marker=shapes[1], facecolor=colors[1], edgecolor=edgecolors[1],
                      label=sample_labels[1], lw=2)
handles = [s1, s2]
_, labels = ax_s[-1].get_legend_handles_labels()

ax_s[-1].legend(handles=handles, labels=labels, fontsize=17,
                handler_map={tuple: HandlerTuple(None)})
fig_s.subplots_adjust(wspace=0.23)
fig_s.show()
plt.pause(1) 
```

```python
fig_l, ax_l = plt.subplots(1, len(title_l0), figsize=(4.4 * len(title_l0), 4), clear=True)
for idx_cal, name in enumerate(title_l):
    func = []
    ax_l[idx_cal].set_title(title_l0[idx_cal], x=.5, y=1.04, fontsize=23)
    ax_l[idx_cal].set_xlabel(r'$R\mathrm{/kpc}$')
    xmin, xmax = 10, 10
    for i in range(2):
        mask = split[1].data[split_type[split_num]] % 2 == i
        ifus, idx, _ = np.intersect1d(parameters[1].data['plateifu'], split[1].data['plateifu'][mask],
                                      return_indices=True)
        _, idx_re, idx0 = np.intersect1d(sga_mass['plateifu'], ifus, return_indices=True)
        re = sga_mass['Re_kpc'][idx_re]
        dx_tmp, dy_tmp = [], []
        for j in range(3):
            # r_tmp=0.5 * j + i * 0.04
            r_tmp = np.nanmedian(parameters[1].data[rnames[j]][idx])
            re_tmp = np.nanmedian(parameters[1].data[rnames[j]][idx] / re)
            data['r_kpc'][i][j].append(r_tmp)
            data['r_re'][i][j].append(re_tmp)

            # idx_r = np.where(r_median[1].data['plateifu'] == '10513-12701')[0][0]
            if r_tmp < xmin:
                xmin = np.copy(r_tmp)
            if r_tmp > xmax:
                xmax = np.copy(r_tmp)

            ax_l[idx_cal].errorbar([r_tmp], [data[name][i][j]],
                                   # xerr=[np.nanstd(parameters[1].data[rnames[j]][idx])],
                                   yerr=[err[name][i][j]],
                                   ecolor=edgecolors[i], linewidth=1.5, capsize=10, capthick=2, linestyle=None,
                                   alpha=1)
            ax_l[idx_cal].scatter([r_tmp], [data[name][i][j]], s=200,
                                  edgecolors=edgecolors[i], facecolors='None', linewidths=1.5,
                                  marker=shapes[i],
                                  alpha=1)
            ax_l[idx_cal].errorbar([r_tmp], [fdata[name][i][j]], xerr=[np.nanstd(parameters[1].data[rnames[j]][idx])],
                                   yerr=[ferr[name][i][j]],
                                   ecolor=edgecolors[i], linewidth=1.5, capsize=10, capthick=2, linestyle=None,
                                   alpha=1)
            ax_l[idx_cal].scatter([r_tmp], [fdata[name][i][j]], s=200,
                                  edgecolors=edgecolors[i], facecolors=colors[i], linewidths=1.5,
                                  marker=shapes[i],
                                  alpha=1)

            dx_tmp.append(r_tmp)
            dy_tmp.append(data[name][i][j][0])

s1 = ax_l[-1].scatter([], [], s=200, marker=shapes[0], facecolor=colors[0], edgecolor=edgecolors[0],
                      label=sample_labels[0], lw=2)
s2 = ax_l[-1].scatter([], [], s=200, marker=shapes[1], facecolor=colors[1], edgecolor=edgecolors[1],
                      label=sample_labels[1], lw=2)
s3 = ax_l[-1].scatter([], [], s=100, marker=shapes[0], facecolor='None', edgecolor='black',
                      label='Kroupa IMF',
                      lw=2)
s4 = ax_l[-1].scatter([], [], s=100, marker=shapes[1], facecolor='None', edgecolor='black',
                      lw=2)

handles = [s1, s2, (s3, s4)]
_, labels = ax_l[-1].get_legend_handles_labels()

ax_l[-1].legend(handles=handles, labels=labels, fontsize=17,
                handler_map={tuple: HandlerTuple(None)})
fig_l.subplots_adjust(wspace=0.23)
fig_l.show()
```
